```python
# Cell 5 Question2 : How to clean the dataset by naming the missing column name
# df.info() shows the firts column as Unamed.
# renaming this column and updating the dataframe:
df.rename(columns={"Unnamed: 0": "vehicle"}, inplace=True)
df.head()
```

```python
# Cell 6 Question3
# Create a dataframe with only the required columns.

# vehicle_stats_df = df[["mpg", "hp", "wt", "cyl"]]         # This is only a comment to check in future stats. panda

vehicle_stats_df = df[["mpg", "hp", "wt", "cyl"]].copy()  # independent dataframe multiple columns using a list ([[ ]]).

print(type(vehicle_stats_df),'\n' )
print("New subset: vehicle_stats_df\n")

display(vehicle_stats_df.head())         # used display otherwise it does not head() would not print anything to the terminal

print("\n===== SHAPE / ATTRIBUTE =====") # Tuple, printed normally
print("Rows:", vehicle_stats_df.shape[0])
print("Columns:", vehicle_stats_df.shape[1], "\n")
```

```python
# Cell 7 Question4 rename the columns with full names
vehicle_stats_df.rename(
    columns={
        "mpg": "Mileage",
        "hp": "Horsepower",
        "wt": "Weight",
        "cyl": "Cylinders"
    },
    inplace=True
)

vehicle_stats_df.head()
```

```python
# Cell 9 Question5
# Summary statistics for the selected columns.

vehicle_stats_df.describe()
```

```python
# Cell 10 Question5
# Calculate the average fuel efficiency.

average_Mileage = vehicle_stats_df["Mileage"].mean()

print(f"Average MPG: {average_Mileage:.2f}")
```

```python
# Cell 10.1 Using df.loc
# Calculate the average fuel efficiency using loc
average_mpg_4cyl = vehicle_stats_df.loc[
    vehicle_stats_df["Cylinders"] == 4,
    "Mileage"
].mean()
print(f"Average mpg_4cyl:  {average_mpg_4cyl:.2f}")
```

```python
# Cell 11 Question 6 Filter for all vehicles using 4 cylinders
# Filter only 4-cylinder vehicles using loc.

four_cylinder_df = vehicle_stats_df.loc[
    vehicle_stats_df["Cylinders"] == 4
]

display(four_cylinder_df.head())
```

```python
# Cell 11.1
# Create a dataframe containing only 4-cylinder vehicles.

four_cylinder_df = vehicle_stats_df[
    vehicle_stats_df["Cylinders"] == 4
]
print(four_cylinder_df.head())
# display(four_cylinder_df.head())  # keeping this line only as a reference to collab and juniper lab
print('\n')
print("\n===== DATAFRAME INFO =====\n")
four_cylinder_df.info()
```

```python
# Cell 12
# Compare fuel efficiency of 4-cylinder vehicles.

average_mpg_4cyl = four_cylinder_df["Mileage"].mean()

print(f"Average MPG for 4-cylinder vehicles: {average_mpg_4cyl:.2f}")
print()
print(type(four_cylinder_df))
```

```python
# Cell Opitonal
# Add a new test vehicle row to the original dataframe.

new_vehicle = {
    "vehicle": "Test Car",
    "mpg": 35.0,
    "cyl": 4,
    "disp": 120.0,
    "hp": 95,
    "drat": 3.90,
    "wt": 2.10,
    "qsec": 19.0,
    "vs": 1,
    "am": 1,
    "gear": 5,
    "carb": 2
}

# Add the new row at the bottom of the dataframe.
df.loc[len(df)] = new_vehicle

print("New vehicle added.\n")
display(df.tail())
```

```python
# Cell Optional
# Verify that the dataframe now contains one additional row.

print("Updated Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Cell Optional
# Remove the temporary test vehicle using its unique vehicle name.

df = df[df["vehicle"] != "Test Car"]

print("Test Car removed.\n")
display(df.tail())
```

```python
# Cell Optional
# Create a new column that measures fuel efficiency relative
# to vehicle weight. Larger values indicate better fuel
# efficiency for the vehicle's weight.

df["efficiency_weight_ratio"] = df["mpg"] / df["wt"]

print(
    df[["vehicle", "mpg", "wt", "efficiency_weight_ratio"]]
    .head()
)
```

```python
# Cell Optional
# Display summary statistics for the newly created
# efficiency_weight_ratio column.

print(df["efficiency_weight_ratio"].describe())
```

```python
# Cell Optional
# Remove the temporary efficiency_weight_ratio column.

df.drop("efficiency_weight_ratio", axis=1, inplace=True)

print("efficiency_weight_ratio column removed.\n")
print(df.columns)
```
