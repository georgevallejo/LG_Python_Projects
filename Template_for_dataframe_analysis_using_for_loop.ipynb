```python
# CELL 1
import pandas as pd
df = pd.read_csv(
    'https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/Salaries.csv'
)
# check datatype
print(type(df))
```

```python
# Cell 0 — Initial Dataset Overview
# Purpose:
# Configure Pandas display settings for clean numeric formatting and generate a
# comprehensive initial exploratory snapshot of the dataset. This cell prints:
# - head
# - shape
# - numeric describe
# - object describe
# - count
# - unique values
# - missing values
# - data types
# - random sample
# - correlation matrix
# - value counts (rank, sex)
# - salary summary
# - info
# This provides a strong initial analysis and verifies data integrity before deeper work.
```

```python
# Main Cell 1 — Initial Dataset Overview
# ------------------------------------------------------------
# Cell Purpose:
# Configure Pandas display settings for clean numeric formatting
# and generate an initial exploratory overview of the dataset.
# This cell prints the first rows, dataset shape, descriptive
# statistics, unique counts, missing values, data types, duplicate
# row count, and a random sample. It provides a quick integrity
# check and a structured snapshot of the dataset before deeper
# analysis.
# ------------------------------------------------------------

# Import Pandas and display utilities
import pandas as pd
from IPython.display import display

# Set Pandas display width so tables don't wrap
pd.set_option('display.width', 200)

# Show all columns without truncation
pd.set_option('display.max_columns', None)

# Format all floating‑point numbers to 2 decimal places
pd.set_option('display.float_format', '{:.2f}'.format)

# Build a dictionary of summary sections for initial analysis
sections = {
    # First 5 rows of the dataset
    "HEAD": df.head(),

    # Number of rows and columns
    "SHAPE": df.shape,

    # Summary statistics for numeric columns
    "DESCRIBE NUMERIC": df.describe(),

    # Summary statistics for object (string/categorical) columns
    "DESCRIBE OBJECT": df.describe(include='object'),

    # Count of non‑missing values per column
    "COUNT": df.count(),

    # Number of unique values per column
    "UNIQUE VALUES": df.nunique(),

    # Missing values per column
    "MISSING VALUES": df.isnull().sum(),

    # Data types for each column
    "DATA TYPES": df.dtypes,

    # Number of duplicated rows in the dataset
    "DUPLICATE ROWS": df.duplicated().sum(),

    # Random sample of 5 rows for variety
    "SAMPLE (5 rows)": df.sample(5)
}

# Loop through each section and print its contents
for title, content in sections.items():
    print(f"===== {title} =====")
    print(content)
    print("\n")

# df.info() prints directly, so it must be handled separately
print("===== INFO =====")
df.info()
print("\n")
```

```python
# Main Cell 2 — Additional Exploratory Analysis sample
# ------------------------------------------------------------
# This cell performs deeper structural checks on the dataset.
# These functions are optional but useful for understanding:
# - numeric relationships (correlation matrix)
# - categorical distributions (value counts)
# - detailed numeric summaries for specific columns
# ------------------------------------------------------------

# Correlation matrix for numeric columns only
# Helps identify relationships between numeric variables.
print("===== CORRELATION MATRIX =====")
print(df.corr(numeric_only=True))
print("\n")

# Value counts for a categorical column (example: rank)
# Shows how many times each category appears.
# Replace 'rank' with any categorical column in your dataset.
print("===== VALUE COUNTS (rank) =====")
print(df['rank'].value_counts())
print("\n")

# Value counts for another categorical column (example: sex)
# Replace 'sex' with any categorical column in your dataset.
print("===== VALUE COUNTS (sex) =====")
print(df['sex'].value_counts())
print("\n")

# Detailed numeric summary for a specific column (example: salary)
# Replace 'salary' with any numeric column in your dataset.
print("===== SUMMARY OF SALARY =====")
print(df['salary'].describe())
print("\n")
```

```python
# Cell 1.1 — Initial Dataset Overview
# ------------------------------------------------------------
# Cell Purpose:
# Configure Pandas display settings for clean numeric formatting
# and generate an initial exploratory overview of the dataset.
# This cell prints the first rows, dataset shape, descriptive
# statistics, unique counts, missing values, data types, duplicate
# row count, and a random sample. It provides a quick integrity
# check and a structured snapshot of the dataset before deeper
# analysis.
# ------------------------------------------------------------

# Import Pandas and display utilities
import pandas as pd
from IPython.display import display

# Set Pandas display width so tables don't wrap
pd.set_option('display.width', 200)

# Show all columns without truncation
pd.set_option('display.max_columns', None)

# Format all floating‑point numbers to 2 decimal places
pd.set_option('display.float_format', '{:.2f}'.format)

# Build a dictionary of summary sections for initial analysis
sections = {
    "HEAD": df.head(),                                 # First 5 rows
    "SHAPE": df.shape,                                 # (rows, columns)
    "COLUMN NAMES": df.columns,                        # Column index
    "INDEX": df.index,                                 # Index structure
    "DESCRIBE NUMERIC": df.describe(),                 # Numeric summary
    "DESCRIBE OBJECT": df.describe(include='object'),  # Categorical summary
    "DESCRIBE ALL": df.describe(include='all'),        # Mixed-type summary
    "COUNT": df.count(),                               # Non-missing count
    "UNIQUE VALUES": df.nunique(),                     # Unique values
    "MISSING VALUES": df.isnull().sum(),               # Missing count (FIXED)
    "MISSING VALUES (%)": (df.isnull().mean() * 100).round(2),  # Missing %
    "DATA TYPES": df.dtypes,                           # Column dtypes
    "DUPLICATE ROWS": df.duplicated().sum(),           # Duplicate count
    "SAMPLE (5 rows)": df.sample(5),                   # Random sample
    "MEMORY USAGE": df.memory_usage(deep=True)         # Memory footprint
}

# Loop through each section and print its contents
for title, content in sections.items():
    print(f"===== {title} =====")
    print(content)
    print("\n")

# df.info() prints directly, so it must be handled separately
print("===== INFO =====")
df.info()
print("\n")
```

```python
# Cell 1.2 — Automatic Histograms for Numeric Columns
# ------------------------------------------------------------
# Purpose:
# Generate histograms for all numeric columns to visualize
# distributions, skewness, and potential outliers.
# ------------------------------------------------------------

import matplotlib.pyplot as plt

numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col].dropna(), bins=30, color='steelblue', edgecolor='black')
    plt.title(f'Histogram — {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
```

```python
# Cell 1.3 — Boxplots using Pandas + Matplotlib (no seaborn)
# ------------------------------------------------------------
# Purpose:
# Generate boxplots for all numeric columns using pandas wrappers.
# ------------------------------------------------------------

import matplotlib.pyplot as plt

numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    df.boxplot(column=col)
    plt.title(f'Boxplot — {col}')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
```

```python
# Cell 1.3 — Automatic Boxplots for Numeric Columns
# ------------------------------------------------------------
# Purpose:
# Generate boxplots for all numeric columns to visualize
# spread, quartiles, and outliers.
# ------------------------------------------------------------

import seaborn as sns

numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col], color='lightgreen')
    plt.title(f'Boxplot — {col}')
    plt.xlabel(col)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
```

```python
# Cell 1.4 — Correlation Heatmap
# ------------------------------------------------------------
# Purpose:
# Display a correlation heatmap for all numeric columns.
# Helps identify relationships and multicollinearity.
# ------------------------------------------------------------

numeric_df = df.select_dtypes(include='number')

if numeric_df.shape[1] > 1:
    plt.figure(figsize=(10,8))
    sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
    plt.title('Correlation Heatmap')
    plt.show()
else:
    print("Correlation Heatmap skipped — not enough numeric columns.\n")
```

```python
# Cell 1.5 — Categorical Bar Charts
# ------------------------------------------------------------
# Purpose:
# Generate bar charts for all categorical (object) columns.
# Helps visualize category frequency distributions.
# ------------------------------------------------------------

cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    plt.figure(figsize=(8,4))
    df[col].value_counts().plot(kind='bar', color='coral', edgecolor='black')
    plt.title(f'Bar Chart — {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
```

```python
# Boxplot Comparison: Pandas vs Seaborn
# ------------------------------------------------------------
# Purpose:
# Show the visual difference between pandas and seaborn boxplots
# using the same numeric column, side-by-side.
# ------------------------------------------------------------

import matplotlib.pyplot as plt
import seaborn as sns

# Pick the first numeric column for comparison
numeric_cols = df.select_dtypes(include='number').columns

if len(numeric_cols) == 0:
    print("No numeric columns available for boxplot comparison.")
else:
    col = numeric_cols[0]  # use the first numeric column

    plt.figure(figsize=(12,5))

    # --- Pandas boxplot ---
    plt.subplot(1, 2, 1)
    df.boxplot(column=col)
    plt.title(f"Pandas Boxplot — {col}")
    plt.grid(True, linestyle='--', alpha=0.5)

    # --- Seaborn boxplot ---
    plt.subplot(1, 2, 2)
    sns.boxplot(x=df[col], color='lightgreen')
    plt.title(f"Seaborn Boxplot — {col}")
    plt.grid(True, linestyle='--', alpha=0.5)

    plt.tight_layout()
    plt.show()
```
